Utilizando Pandas:

- Normalizar las de fechas con el formato 'YYYY-MM-DD', en caso de fechas inválidas completar con la fecha `1932-01-01`. Mostrar las 10 primeras fechas.

- Normalizar los horas al formato de 24hs, las horas inválidas se procesaran como `00:00`. Mostrar las 10 primeras horas.

- Normalizar las ubicaciones. Quitar caracteres especiales y pasar todo a mayúsculas. Mostrar las 10 primeras ubicaciones.

- Eliminar las filas que posean valores relevantes para las multas y que se encuentren vacios. Mostrar la cantidad de registros eliminados y cuales son las columnas que se observaron utilizando la leyenda

- Crear la columna `exceso_velocidad_real` y calcularla. El exceso de velocidad real es la diferencia entre velocidad_registrada y la velocidad_maxima. Mostrar las primeras 10 patentes y su exceso de velocidad real.

- Crear la columna `exceso_velocidad` y calcularla. El exceso de velocidad es la diferencia entre velocidad_registrada y la velocidad_maxima más un 5 porciento de la velocidad máxima. Mostrar las primeras 10 patentes y su exceso de velocidad.

- Eliminar las filas que no cometieron infracciones según el exceso de velocidad.Mostrar cuantas filas se eliminaron con la leyenda
```

```

- Guardar el dataset limpio en `urban_flow/data/interim/speeding_fines.csv`

````python
# proceso de limpieza y normalizacion
separacion("fecha")
def normalizar_fecha(fecha):
    try:
        return pd.to_datetime(fecha).strftime('%Y-%m-%d')
    except:
        return "1932-01-01"

df['fecha'] = df['fecha'].apply(normalizar_fecha)
fecha = df['fecha'].head(10)
print(fecha)

#hora
separacion("hora")
def normalizar_hora(hora):
    try:
        return pd.to_datetime(hora).strftime('%H:%M')
    except:
        return "00:00"

df['hora'] = df['hora'].apply(normalizar_hora)
hora = df['hora'].head(10)
print(hora)

#ubicacion
import re
separacion("ubicacion")

def limpiar_texto(texto):
    if pd.isna(texto):
        return texto
    texto = re.sub(r'[^A-Za-z0-9 ]', '', texto)
    return texto.upper()

df['ubicacion'] = df['ubicacion'].apply(limpiar_texto)
ubicacion = df['ubicacion'].head(10)
print(ubicacion)

#limpieza de patentes
separacion("patentes")
def limpiar_patente(p):
    if pd.isna(p):
        return pd.NA

    p = re.sub(r'[^A-Za-z0-9]', '', p).upper()
    if len(p) < 6:
        return pd.NA

    return p

df['patente'] = df['patente'].apply(limpiar_patente)
patentes = df['patente'].tail(10)
print(patentes)

#Limpiar datos
separacion("limpiar datos")
columnas = ['patente', 'velocidad_registrada', 'velocidad_maxima']

antes = len(df)

df = df.dropna(subset=columnas)

despues = len(df)

print(f"Se eliminaron {antes - despues} filas.")
print("Las columnas observadas son:")
for col in columnas:
    print(f" - {col}")

#exceso de velocidad
separacion("exceso de velocidad")
df['exceso_velocidad_real'] = (
    df['velocidad_registrada'] - df['velocidad_maxima']
)

exceso_vel = df[['patente', 'exceso_velocidad_real']].head(10)
print(exceso_vel)


#tolerancia
separacion("tolerancia")
df['exceso_velocidad'] = (
    df['velocidad_registrada'] - (df['velocidad_maxima'] * 1.05)
)

tolerancia = df[['patente', 'exceso_velocidad']].head(10)
print(tolerancia)

#filtrar infracciones
separacion("filtrar infracciones")

antes = len(df)

df = df[df['exceso_velocidad'] > 0]

despues = len(df)

print(f"Se eliminaron {antes - despues} filas.")

#sin infracciones
separacion("sin infracciones")
Q1 = df['velocidad_registrada'].quantile(0.25)
Q3 = df['velocidad_registrada'].quantile(0.75)
IQR = Q3 - Q1

antes = len(df)

df = df[
    (df['velocidad_registrada'] >= Q1 - 1.5 * IQR) &
    (df['velocidad_registrada'] <= Q3 + 1.5 * IQR)
]

despues = len(df)

print(f"Se eliminaron {antes - despues} filas.")
print("Las columnas observadas son:")
print(" - velocidad_registrada")

#guardar datos
separacion("guardar datos")
os.makedirs("urban_flow/data/interim", exist_ok=True)
df.to_csv("urban_flow/data/interim/speeding_fines.csv", index=False)
````

## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
separacion("Ejercicio 4")
class FineAnalyzer:
    def __init__(self, df):
        self.df = df.copy()

    def top_patentes(self) -> pd.DataFrame:
        """Muestra el top 5 de patentes"""
        result = self.df['patente'].value_counts().head(5)
        result = result.reset_index()
        result.columns = ['patente', 'cantidad']
        result.index += 1
        print(result)
        return result

    def top_horas(self) -> pd.DataFrame:
        """Muestra la cantidad de hs"""
        result = self.df['hora'].value_counts().head(5)
        result = result.reset_index()
        result.columns = ['hora', 'cantidad']
        result.index += 1
        print( result)
        return result

    def exceso_promedio(self) -> float:
        """Muestra el promedio de exceso de velocidad"""
        result = float(self.df['exceso_velocidad'].mean())
        print( result)
        return result

    def exceso_real_promedio(self) -> float:
        """Muestra el promedio de exceso de velocidad real"""
        result = float(self.df['exceso_velocidad_real'].mean())
        print( result)
        return result

    def multas_por_ubicacion(self) -> pd.DataFrame:
        """Muestra la cantidad de multas por ubicacion"""
        result = self.df['ubicacion'].value_counts().reset_index()
        result.columns = ['ubicacion', 'cantidad']
        print( result)
        return result

analizador = FineAnalyzer(df)
analizador.top_patentes()
analizador.top_horas()
analizador.exceso_promedio()
analizador.exceso_real_promedio()
analizador.multas_por_ubicacion()
````
